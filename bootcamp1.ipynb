```python
marks = [45, 60, 72, 38, 90]

total = 0

for i in range(len(marks) - 1):
    total += marks[i]

average = total / len(marks)

if average > 50:
    print("Class passed")
else:
    print("Class failed")

print("Average:", average)
```

```python
import pandas as pd
```

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106],
    "attendance_percent": [92, 67, 81, 45, 74, 88],
    "assignment_score": [18, 12, 15, 8, 14, 19],
    "quiz_score": [72, 48, 65, 30, 55, 80],
    "lab_completed": [True, False, True, False, True, True]
}

df = pd.DataFrame(data)

df
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['total_score'] = df['assignment_score'] + df['quiz_score']
df
```

```python
df['eligible'] = (
    (df['attendance_percent'] >= 75) &
    (df['total_score'] >= 70) &
    (df['lab_completed'])
)
df
```

```python
eligible_students = df[df["eligible"] == True]
eligible_students
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(df["student_id"], df["attendance_percent"], label="Attendance %")
plt.bar(df["student_id"], df["quiz_score"], label="Quiz Score", alpha=0.7)
plt.xlabel("Student ID")
plt.ylabel("Values")
plt.title("Attendance vs Quiz Score")
plt.legend()
plt.show()
```

```python

plt.scatter(df["attendance_percent"], df["quiz_score"], c=df["assignment_score"], cmap="viridis", s=100)
plt.xlabel("Attendance %")
plt.ylabel("Quiz Score")
plt.title("Scatter: Attendance vs Quiz (color=Assignment)")
plt.colorbar(label="Assignment Score")
plt.show()
```

```python
plt.plot(df["student_id"], df["total_score"], marker="o", linestyle="--", color="purple")
plt.xlabel("Student ID")
plt.ylabel("Total Score")
plt.title("Total Performance Score Trend")
plt.show()
```
